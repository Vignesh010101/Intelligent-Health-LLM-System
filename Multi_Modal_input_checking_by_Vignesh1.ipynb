```python
import openai
import os
import fitz  # PyMuPDF for PDF processing
import base64  # For encoding images

# Set your API key
openai.api_key = "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"


def extract_mri_text_from_pdf(pdf_path):
    """
    Extracts MRI-related text from a medical PDF document (e.g., MRI report).
    Uses PyMuPDF (fitz) for efficient text extraction.
    """
    try:
        if not os.path.exists(pdf_path):
            return "Error: File not found."

        text = ""
        with fitz.open(pdf_path) as doc:
            for page in doc:
                page_text = page.get_text("text")
                if "MRI" in page_text or "magnetic resonance imaging" in page_text.lower():
                    text += page_text + "\n"

        return text.strip() if text.strip() else "Error: No MRI-related text detected in the PDF."

    except Exception as e:
        return f"Error extracting text from PDF: {str(e)}"

def encode_image(image_path):
    """
    Converts an image (JPG, PNG) into base64 format for API compatibility.
    """
    try:
        with open(image_path, "rb") as img_file:
            return base64.b64encode(img_file.read()).decode("utf-8")
    except Exception as e:
        return f"Error encoding image: {str(e)}"

def analyze_medical_document(file_path):
    """
    Sends the image or extracted MRI text to LLM model for analysis.
    Supports multimodal input (image or text).
    """
    try:
        if not os.path.exists(file_path):
            return "Error: The file does not exist. Check the path."

        if file_path.lower().endswith((".png", ".jpg", ".jpeg")):
            # Convert image to Base64
            image_data = encode_image(file_path)
            if "Error" in image_data:
                return image_data

            # OpenAI Vision Model Inference
            response = openai.ChatCompletion.create(
                model="gpt-4-turbo",
                messages=[
                    {"role": "system", "content": "You are a medical expert analyzing MRI scans and reports."},
                    {"role": "user", "content": [
                        {"type": "text", "text": "Analyze this MRI scan and provide detailed medical insights."},
                        {"type": "image_url", "image_url": {"url": f"data:image/jpeg;base64,{image_data}"}}
                    ]}
                ],
                temperature=0.9,
                max_tokens=1250
            )


            return response["choices"][0]["message"]["content"]

        elif file_path.lower().endswith(".pdf"):
            extracted_text = extract_mri_text_from_pdf(file_path)
            if "Error" in extracted_text:
                return extracted_text  # Return error if MRI extraction fails

            # OpenAI Text-based Analysis
            response = openai.ChatCompletion.create(
                model="gpt-4-turbo",
                messages=[
                    {"role": "system", "content": "You are a medical expert analyzing MRI reports."},
                    {"role": "user", "content": f"Extracted MRI text from document:\n{extracted_text}\n\n"
                                                 "Analyze this MRI report and provide a detailed medical recommendation, "
                                                 "including any potential conditions, observations, and next steps."}
                ],
                max_tokens=500
            )
            return response["choices"][0]["message"]["content"]

        else:
            return "Unsupported file type. Please provide a JPG, PNG, or PDF."

    except Exception as e:
        return f"Error processing the document: {str(e)}"

def main():
    file_path = input("Enter the path to your medical file (image or PDF): ").strip()

    result = analyze_medical_document(file_path)

    print("\n📄 **Medical Analysis Result:**")
    print(result)

if __name__ == "__main__":
    main()
```

```python
import openai
import os
import IPython
import torchaudio
import tempfile
from google.colab import output
from IPython.display import Audio, display

# Set API Key
openai.api_key = "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"

def record_audio_colab():
    """
    Records an audio clip in Colab using JavaScript.
    """
    print("🎤 Recording... Speak now!")
    audio_path = tempfile.NamedTemporaryFile(suffix=".wav").name
    output.eval_js('''
        async function recordAudio() {
            const stream = await navigator.mediaDevices.getUserMedia({ audio: true });
            const recorder = new MediaRecorder(stream);
            let chunks = [];
            recorder.ondataavailable = e => chunks.push(e.data);
            recorder.start();
            await new Promise(resolve => setTimeout(resolve, 5000)); // Record for 5 sec
            recorder.stop();
            await new Promise(resolve => recorder.onstop = resolve);
            let blob = new Blob(chunks, { type: 'audio/wav' });
            let reader = new FileReader();
            reader.readAsDataURL(blob);
            await new Promise(resolve => reader.onloadend = resolve);
            return reader.result.split(',')[1];
        }
    ''')
    audio_bytes = output.eval_js('recordAudio()')
    with open(audio_path, "wb") as f:
        f.write(IPython.utils.decodebytes(audio_bytes.encode()))
    return audio_path

def transcribe_audio_with_whisper(audio_path):
    """
    Transcribes speech using Whisper.
    """
    try:
        with open(audio_path, "rb") as audio_file:
            response = openai.Audio.transcribe(model="whisper-1", file=audio_file)
        return response["text"]
    except Exception as e:
        return f"Error transcribing audio: {str(e)}"

def get_medical_response(transcribed_text):
    """
    Uses Models to generate medical insights based on transcribed text.
    """
    prompt = f"Patient's spoken input: {transcribed_text}\nProvide detailed, professional medical advice based on this input."

    try:
        response = openai.ChatCompletion.create(
            model="gpt-4-turbo",
            messages=[{"role": "user", "content": prompt}],
            temperature=0.7,
            max_tokens=800
        )
        return response["choices"][0]["message"]["content"]
    except Exception as e:
        return f"Error generating medical response: {str(e)}"

def main():
    audio_path = record_audio_colab()
    print("\n🔄 Transcribing...")
    transcribed_text = transcribe_audio_with_whisper(audio_path)

    print("\n📝 Transcribed Text:")
    print(transcribed_text)

    if transcribed_text and not transcribed_text.startswith("Error"):
        result = get_medical_response(transcribed_text)
        print("\n🩺 Medical Analysis from Voice Input:")
        print(result)
    else:
        print("⚠️ Failed to transcribe audio.")

if __name__ == "__main__":
    main()
```
